# Optimizing Lead Quality

> If the advertiser says they will increase our CPL by 20\% (i.e., \$30 to \$33) if we increase our lead quality by 20\% (i.e., from 8.0\% to 9.6\%), do we see any opportunities to do that here? What kinds of things could we do?

\subsection{The Target}

The advertiser's requirement is to increase the overall lead quality rate from
\[
8.0\%
\quad\text{to}\quad
9.6\%.
\]

This is a \(20\%\) relative increase, although it corresponds to only a
\[
9.6\%-8.0\%=1.6
\]
percentage-point increase in the absolute closure rate.

The current cost per closed lead, using a \$30 CPL, is approximately
\[
\frac{\$30}{0.08}=\$375.
\]

If the target is achieved and the advertiser pays \$33 per lead, the corresponding cost per closed lead becomes
\[
\frac{\$33}{0.096}=\$343.75.
\]

Thus, achieving the target would actually improve the economics of the traffic from the advertiser's perspective:
\[
\frac{343.75}{375}-1\approx -8.3\%.
\]

In other words, the advertiser is willing to pay \(10\%\) more for each closed lead while receiving \(20\%\) more closed leads per unit of traffic. This creates a meaningful incentive to optimize quality rather than simply maximize lead volume.

\subsection{There Is Sufficient Variation in the Data}

The dataset provides substantial evidence that a \(1.6\)-percentage-point improvement is achievable in principle.

The overall closure rate is approximately \(8.1\%\), but closure rates vary considerably across acquisition and advertising segments. For example:

This suggests that the most obvious opportunity is not to redesign the entire acquisition funnel. It is to \emph{shift the composition of traffic toward segments that already demonstrate higher downstream quality}.

\subsection{Opportunity 1: Shift Traffic Toward Higher-Intent Ad Groups}

The clearest opportunity is the large variation across \texttt{AdGroupFamily}.

Debt Settlement, Stop Collections, and Consolidate all have closure rates above \(12\%\), while Student Debt and Loan Default are around \(1\%\).

This suggests reducing exposure to very low-quality intent categories and reallocating some of the corresponding acquisition budget toward higher-quality intent categories.

For example, suppose the current portfolio contains a proportion \(w_L\) of leads from low-quality segments with closure rate \(q_L\), and a proportion \(w_H\) from high-quality segments with closure rate \(q_H\). The aggregate closure rate is

\[
Q=w_Hq_H+(1-w_H)q_L.
\]

If \(q_H>Q>q_L\), then increasing \(w_H\) necessarily increases the aggregate quality rate.

The important point is that the required improvement is relatively modest. Moving from \(8.0\%\) to \(9.6\%\) does not require all traffic to resemble the best-performing segments. It requires a relatively small change in the mix.

\subsection{Opportunity 2: Reduce Low-Quality Student-Debt Traffic}

The Student Debt segment deserves particular attention because it is both:

\begin{enumerate}
	\item relatively large, with approximately 392 leads; and
	\item extremely low quality, with a closure rate of approximately \(1.0\%\).
\end{enumerate}

This is much more actionable than a small segment with an extreme rate.

If a substantial amount of traffic is currently being purchased against student-debt-related intent, there is an opportunity to reduce bids, reduce exposure, or modify the targeting of these keywords and ad groups.

Importantly, the goal should not necessarily be to eliminate Student Debt traffic completely. The appropriate strategy would be to determine whether the segment can be improved through tighter keyword targeting or whether its economics are intrinsically poor.

\subsection{Opportunity 3: Reallocate Traffic from the DRI Campaign}

The DRI campaign is another particularly important target because of its scale.

It accounts for approximately 1,496 leads but has a closure rate of only about \(5.9\%\). By contrast, the DHT campaign produces a \(14.5\%\) closure rate.

This creates a potentially attractive reallocation opportunity.

The important distinction is:

\[
\text{high volume} \neq \text{high value}.
\]

Even if DRI is currently responsible for a large proportion of total leads, its relatively low downstream quality means that some of its traffic may be worth redirecting toward higher-quality campaigns.

A useful operational experiment would therefore be to move a controlled fraction of DRI budget into DHT, FS, or other high-quality campaigns and measure the resulting change in both lead volume and closure rate.

\subsection{Opportunity 4: Optimize for Search Intent Rather Than Raw Lead Volume}

The ad-group results suggest that the underlying problem expressed by the customer is more important than simply generating a form submission.

The strongest segments are associated with explicit intent such as:

\begin{itemize}
	\item debt settlement;
	\item debt negotiation;
	\item stopping collections;
	\item consolidation.
\end{itemize}

The weakest segments include:

\begin{itemize}
	\item student debt;
	\item loan default;
	\item lower-payment searches.
\end{itemize}

Therefore, keyword and ad-group bidding should ideally be based on downstream quality rather than solely on cost per lead.

For example, suppose two keyword groups have the same CPL:

\[
\text{Keyword A: } \$30,\quad Q_A=15\%,
\]

and

\[
\text{Keyword B: } \$30,\quad Q_B=5\%.
\]

Their cost per closed lead is

\[
\frac{30}{0.15}=\$200
\]

and

\[
\frac{30}{0.05}=\$600,
\]

respectively.

Thus, the cheaper-looking lead source can actually be three times more expensive when evaluated on the outcome that matters.

\subsection{Opportunity 5: Use Applicant Debt Level for Quality-Oriented Targeting}

Debt level provides another potential lever.

The data show that several debt bands have closure rates above the overall average, while the lowest debt bands and the \$100k+ category perform poorly.

This suggests that applicant financial characteristics could potentially be incorporated into bidding or lead-scoring decisions.

However, this is a less direct intervention than ad targeting. The advertiser generally cannot know a person's debt level before the person submits the lead. Consequently, debt level is more useful for:

\begin{enumerate}
	\item estimating expected lead quality;
	\item developing a predictive lead score;
	\item optimizing the mix of traffic that reaches the form;
	\item and evaluating whether particular acquisition sources disproportionately attract desirable applicants.
\end{enumerate}

It is less useful as a direct pre-click targeting variable unless the advertising platform provides an appropriate and permissible proxy.

\subsection{Opportunity 6: Do Not Prioritize Physical Placement}

The data do not provide convincing evidence that physical ad placement is an effective quality lever.

Top Right--300$\times$250 has an observed closure rate of approximately \(9.6\%\), compared with approximately \(8.0\%\) for Top Left--302$\times$252. However, the difference is not statistically significant.

Consequently, reallocating traffic solely on the basis of physical placement would not be my first recommendation.

The more promising strategy is to determine \emph{which advertisements and audiences are being delivered through each placement}. The apparent placement difference may largely reflect differences in the campaigns and ad groups assigned to each placement.

\subsection{Opportunity 7: Test Creative, but Treat It as a Secondary Lever}

There are visible differences between individual creatives, but the evidence is weaker than for ad-group family and acquisition source.

For example, some creatives have observed closure rates above \(10\%\), while others are substantially lower. However, several of the extreme values come from relatively small samples.

Creative testing should therefore be conducted as an experiment rather than by simply selecting the creative with the highest historical rate.

A suitable test would hold the targeting and acquisition conditions approximately constant and randomize users between alternative creatives. The resulting difference in closure rates could then be interpreted much more confidently.

\subsection{A Practical Optimization Strategy}

I would therefore pursue the following strategy, in roughly this order:

\begin{enumerate}
	\item \textbf{Reduce or constrain the lowest-quality intent segments.}
	Student Debt and Loan Default are the clearest candidates, given their extremely low closure rates.
	
	\item \textbf{Reallocate some traffic toward high-quality intent.}
	Increase exposure to Debt Settlement, Stop Collections, Consolidate, and Holding Tank-type traffic, subject to available volume and acceptable acquisition costs.
	
	\item \textbf{Shift campaign allocation based on downstream outcomes.}
	DRI generates substantial volume but relatively low-quality leads. A controlled reallocation toward DHT and FS is a natural experiment.
	
	\item \textbf{Introduce quality-adjusted bidding.}
	Instead of optimizing for CPL alone, use an estimate of
	\[
	P(\text{Closed}\mid X)
	\]
	to determine the maximum economically sensible CPL for each segment.
	
	\item \textbf{Run controlled creative experiments.}
	Use the historical creative differences to identify candidates for testing, but do not infer causality from the observational data.
	
	\item \textbf{Do not prioritize placement or state without additional evidence.}
	The observed differences are not statistically convincing enough to justify major budget changes.
\end{enumerate}

\subsection{Can the 20\% Quality Target Be Achieved?}

Based on the observed heterogeneity in the data, I believe that the target of increasing the closure rate from \(8.0\%\) to \(9.6\%\) is \emph{plausible}.

The reason is that the target requires only a \(1.6\)-percentage-point increase, whereas several large, observable segments already have closure rates well above \(9.6\%\). In particular, the DHT campaign has a closure rate of approximately \(14.5\%\), while Debt Settlement and Stop Collections ad groups have closure rates of approximately \(16.1\%\) and \(13.8\%\), respectively.

However, the existence of high-quality segments does not by itself guarantee that their traffic can be scaled without degradation. These segments may have limited inventory, and their observed quality may partly reflect differences in applicant composition or acquisition source.

The correct next step is therefore a controlled reallocation experiment rather than an immediate wholesale change in budget.

For example, the advertiser could initially shift a modest fraction of spend away from the lowest-quality segments and toward the strongest high-intent segments. The experiment should measure:

\[
\begin{aligned}
	\text{CPL},\qquad
	&\text{Qualification Rate},\\
	\text{Closure Rate},\qquad
	&\text{Cost per Closed Lead}.
\end{aligned}
\]

The target should be considered successful only if the increase in quality persists without an unacceptable increase in acquisition cost or deterioration in lead volume.

\subsection{Economic Interpretation}

The proposed deal is attractive if the quality improvement can be achieved.

At the current rate:

\[
CPL=\$30,\qquad Q=8.0\%,
\]

giving

\[
CPC_{\text{closed}}
=
\frac{30}{0.08}
=
\$375.
\]

At the proposed rate:

\[
CPL=\$33,\qquad Q=9.6\%,
\]

giving

\[
CPC_{\text{closed}}
=
\frac{33}{0.096}
=
\$343.75.
\]

Therefore, if the advertiser's definition of quality is indeed the closure rate, achieving the target would reduce the cost per closed lead by approximately \(8.3\%\), despite the \(10\%\) increase in CPL.

This makes quality optimization economically worthwhile even before considering any additional lifetime value associated with better leads.

\subsection{Conclusion}

There are clear opportunities to improve lead quality from the existing traffic mix. The most promising lever is \textbf{ad/search intent}, followed by \textbf{acquisition partner and campaign allocation}. Applicant debt level provides an additional useful segmentation variable.

The data suggest moving away from low-quality intent such as Student Debt and Loan Default and toward higher-intent segments such as Debt Settlement, Stop Collections, Consolidate, and Holding Tank traffic. The DRI campaign is also a particularly natural candidate for controlled budget reallocation because it combines very high volume with below-average downstream quality.

By contrast, there is currently insufficient evidence to justify optimizing primarily on physical ad placement, state, or individual creative.

Thus, my recommendation would be:

\begin{quote}
	\textbf{Do not try to get 20\% more ``quality'' by making every lead better. Change the composition of the leads being acquired.}
\end{quote}

The dataset contains enough variation to make a \(20\%\) relative improvement in lead quality plausible. The most defensible way to pursue it is through controlled reallocation toward higher-intent traffic, followed by experimentation and a quality-adjusted bidding strategy based on downstream closure probability.
